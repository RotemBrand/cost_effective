```python
import utilities.read_write as rw

data = rw.read_nxjson(r"data/optimize_plot.nxjson")
for graph in data.graph:
    print(str(graph))

import numpy as np
from indexes import edge_probs_by_length, GraphRel
from indexes.probs import Float
from indexes import get_skeleton_graph
def get_mean_edge_p(row):
    # Calculate probabilities for all edges using the row's p_rate
    probs, rate = edge_probs_by_length(row.graph, p=row.p_rate, mode="rate")
    print(rate)
    nx.set_edge_attributes(row.graph, probs, "prob")
    
    # Return the mean of these probabilities
    return np.mean(list(probs.values()))

def get_saidi(row):
    probs, _ = edge_probs_by_length(row.graph, p=row.p_rate, mode="rate")
    probs = {e: Float(p) for e, p in probs.items()}
    gr = GraphRel(row.graph, edges_prob=probs)
    return gr.calc_rel(rel_kinds=["saidi"])["saidi"]

def get_chanis_data(row):
    graph = row.graph
    probs, rate = edge_probs_by_length(row.graph, p=row.p_rate, mode="rate")
    nx.set_edge_attributes(row.graph, probs, "prob")
    strc = get_skeleton_graph(graph)
    
    chains_total_prob = []
    for c, data in strc.edges.items():
        edge_probs = list(nx.get_edge_attributes(data["subgraph"], "prob").values())
        # Calculate the probability that the chain fails (at least one edge fails)
        chain_prob = 1.0 - np.prod(1.0 - np.array(edge_probs))
        chains_total_prob.append(chain_prob)
    mean, std = np.mean(chains_total_prob), np.std(chains_total_prob)
    return pd.Series({'chains_mean': mean, 'chains_std': std})
# data[["chains_mean", "chains_std"]] = data.apply(get_chanis_data, axis=1)


def mean_edge_len(G):
    pos = nx.get_node_attributes(G, "pos")
    return np.mean([np.linalg.norm(np.array(pos[u]) - np.array(pos[v])) for u, v in G.edges])
data['mean_edge_length'] = data['graph'].apply(mean_edge_len)

def std_edge_len(G):
    pos = nx.get_node_attributes(G, "pos")
    return np.std([np.linalg.norm(np.array(pos[u]) - np.array(pos[v])) for u, v in G.edges])
data['sts_edge_length'] = data['graph'].apply(std_edge_len)


# Apply the function to compute the mean 'p' for each graph
data['mean_p'] = data.apply(get_mean_edge_p, axis=1)
data.loc[1:, 'real_saidi'] = data.loc[1:].apply(get_saidi, axis=1)

for graph in data.graph:
    print(str(graph))
data
```

```python
import numpy as np
import pandas as pd
import networkx as nx
from indexes import get_skeleton_graph

def get_physical_chain_lengths(graph: nx.Graph) -> list:
    comps2 = list(nx.k_edge_components(graph, 2))
    graph_copy = graph.copy()
    pos = nx.get_node_attributes(graph, "pos")
    chains_phys_len = []
    
    def extract_phys_len(strc_graph):
        phys_lens = []
        for u, v, data in strc_graph.edges(data=True):
            chain_subgraph = data.get('subgraph')
            if not chain_subgraph:
                continue
            
            # Calculate total physical length of this chain
            length = 0.0
            for edge_u, edge_v in chain_subgraph.edges():
                # Using the original node positions
                p_u = np.array(pos[edge_u])
                p_v = np.array(pos[edge_v])
                length += np.linalg.norm(p_u - p_v)
            phys_lens.append(length)
        return phys_lens

    # 1. Extract chains from within 2-edge connected components
    for comp in comps2:
        if len(comp) > 1:
            strc_comp = get_skeleton_graph(nx.subgraph(graph, comp))
            chains_phys_len.extend(extract_phys_len(strc_comp))
            
            # Contract the component
            comp_list = list(comp)
            for node in comp_list[1:]:
                nx.contracted_nodes(graph_copy, comp_list[0], node, self_loops=False, copy=False)
            
    # 2. Extract external chains after contraction
    strc = get_skeleton_graph(graph_copy)
    chains_phys_len.extend(extract_phys_len(strc))
    
    return chains_phys_len

def calculate_chain_stats(row):
    lengths = get_physical_chain_lengths(row.graph)
    if not lengths:
        return pd.Series({'chain_length_mean': 0.0, 'chain_length_std': 0.0})
    
    return pd.Series({
        'chain_length_mean': np.mean(lengths),
        'chain_length_std': np.std(lengths)
    })

# Apply the function to compute mean and variance for each graph
data[['chain_length_mean', 'chain_length_var']] = data.apply(calculate_chain_stats, axis=1)
data
```
